## Get list of assets per group or organisation or subgroup

```python
path = 'mix_report_template_uk.xlsx'
sub_groups_df = pd.read_excel(path, sheet_name='SubGroups')
sub_groups_df = sub_groups_df[sub_groups_df.report_name.notna()]
parent_groups = sub_groups_df.parentGroupId.unique()
parent_groups = [parent_group.replace('A', '') for parent_group in parent_groups]


assets_res = [pd.DataFrame(mix_api_utilities.get_group_assets(parent_group)) for parent_group in parent_groups]
assets_df = pd.concat(assets_res)
mix_api_utilities.id_to_string(assets_df)


drivers_res = [pd.DataFrame(mix_api_utilities.get_group_drivers(parent_group)) for parent_group in parent_groups]
drivers_df = pd.concat(drivers_res)
mix_api_utilities.id_to_string(drivers_df)
```

```python
assets_df2 = assets_df.merge(sub_groups_df[['GroupId', 'report_name']].rename(columns={'GroupId':'SiteId'}), on='SiteId', how='left')
assets_df2 = assets_df2[assets_df2.report_name.notna()].copy()

drivers_df2 = drivers_df.merge(sub_groups_df[['GroupId', 'report_name']].rename(columns={'GroupId':'SiteId'}), on='SiteId', how='left')
drivers_df2 = drivers_df2[drivers_df2.report_name.notna()].copy()

#drivers_df2[['SiteId', 'DriverId', 'Name', 'report_name']]
#assets_df2.report_name.notna().value_counts()
```
